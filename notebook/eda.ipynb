```python
df.drop(columns=["fuel_system_type"], inplace=True)
df.isnull().sum()
```

```python
df.describe()
```

```python
numerical_features = ["displacement_cc","cylinder_count","engine_cycle","bore_mm","stroke_mm"]
```

```python
for i in numerical_features:
    plt.figure(figsize=(7,3))
    sns.histplot(df[i], kde=True)
    plt.title(f"Distribution of {i}")
    plt.show()
```

```python
# checking outliers using boxplot
for i in numerical_features:
    plt.figure(figsize=(7,3))
    sns.boxplot(df[i])
    plt.title(f"Boxplot of {i}")
    plt.show()
```

```python
corr=df.corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap="coolwarm")
```

```python
categorical_features = [
    "fuel_injection",
    "is_automatic",
    "cooling_Air",
    "cooling_Liquid",
    "cooling_Oil & air",

]
```

```python
# rellation ship  b/w  cc and bore 
plt.figure(figsize=(7,5))

sns.scatterplot(
    data=df,
    x="displacement_cc",
    y="bore_mm"
)

plt.title("Displacement vs Bore")

plt.show()
```

as the cc increases bore size increases 

```python
plt.figure(figsize=(7,5))

sns.scatterplot(
    data=df,
    x="displacement_cc",
    y="stroke_mm"
)

plt.title("Displacement vs Stroke")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.scatterplot(
    data=df,
    x="displacement_cc",
    y="rear_tyre_width"
)

plt.title("Displacement vs Rear Tyre Width")

plt.show()
```
