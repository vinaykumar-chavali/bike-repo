```python
bike_cleaned_df["displacement_cc"] = (
    bike_cleaned_df["Displacement"]
    .str.extract(r"(\d+(?:\.\d+)?)\s*ccm", expand=False)
    .astype(float)
)
```

```python
#engine type: cylinde count
import re

def get_cylinder_count(engine):
    engine = str(engine).lower()

    # Direct numbers: V2, V4, V6, V8
    match = re.search(r'\bv(\d+)\b', engine)
    if match:
        return int(match.group(1))

    # Words
    numbers = {
        "single": 1,
        "twin": 2,
        "three": 3,
        "four": 4,
        "six": 6,
        "eight": 8
    }

    for word, count in numbers.items():
        if word in engine:
            return count

    return np.nan

bike_cleaned_df["cylinder_count"] = bike_cleaned_df["Engine type"].apply(
    get_cylinder_count
)
```

```python
#engine cycle
def get_engine_cycle(engine):
    engine = str(engine).lower()

    if "four-stroke" in engine:
        return 4
    elif "two-stroke" in engine:
        return 2

    return np.nan

bike_cleaned_df["engine_cycle"] = bike_cleaned_df["Engine type"].apply(
    get_engine_cycle
)
```

```python
bike_cleaned_df.head()
```

```python
# bore, stroke 

def extract_bore_stroke(value):
    match = re.search(
        r'(\d+(?:\.\d+)?)\s*x\s*(\d+(?:\.\d+)?)\s*mm',
        str(value),
        re.IGNORECASE
    )

    if match:
        bore = float(match.group(1))
        stroke = float(match.group(2))
        return pd.Series([bore, stroke])

    return pd.Series([np.nan, np.nan])

bike_cleaned_df[["bore_mm", "stroke_mm"]] = bike_cleaned_df["Bore x stroke"].apply(
    extract_bore_stroke
)

```

```python
bike_cleaned_df[["Bore x stroke", "bore_mm", "stroke_mm"]].head(10)
```

```python
#fuel system 
def simplify_fuel_system(value):
    value = str(value).lower()

    if "injection" in value:
        return "Injection"
    
    elif "carburettor" in value:
        return "Carburettor"
    
    else:
        return "Other"

bike_cleaned_df["fuel_system_type"] = bike_cleaned_df["Fuel system"].apply(
    simplify_fuel_system
)
```

```python
bike_cleaned_df["fuel_system_type"].value_counts()
```

```python
bike_cleaned_df["fuel_injection"] = (
    bike_cleaned_df["fuel_system_type"] == "Injection"
).astype(int)
```

```python
#cooling system 

cooling_encoded = pd.get_dummies(
    bike_cleaned_df["Cooling system"],
    prefix="cooling",
    dtype=int
)

bike_cleaned_df = pd.concat([bike_cleaned_df, cooling_encoded], axis=1)


```

```python
# gear box
import re
import numpy as np

def extract_gears(value):
    match = re.search(r'(\d+)-speed', str(value).lower())

    if match:
        gears = int(match.group(1))

        # Treat unrealistic values as missing
        if 1 <= gears <= 8:
            return gears

    return np.nan

bike_cleaned_df["gear_count"] = bike_cleaned_df["Gearbox"].apply(extract_gears)
```

```python
#tyre  profiles
def extract_tyre(value):

    match = re.search(
        r'(\d+(?:\.\d+)?)\s*/\s*(\d+(?:\.\d+)?)\s*[-R]\s*(\d+(?:\.\d+)?)',
        str(value),
        re.IGNORECASE
    )

    if match:
        width = float(match.group(1))
        aspect = float(match.group(2))
        rim = float(match.group(3))

        return pd.Series([width, aspect, rim])

    return pd.Series([np.nan, np.nan, np.nan])
```

```python
bike_cleaned_df[
    ["front_tyre_width", "front_tyre_aspect", "front_tyre_rim"]
] = bike_cleaned_df["Front tire"].apply(extract_tyre)
```
